# E-Commerce Sales Analysis
Complete data analysis workflow

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

df = pd.read_csv('../data/sales_data.csv')
print(f'Data loaded: {len(df)} rows')
```

```python
print(df.head())
print(df.describe())
```

```python
product_revenue = df.groupby('product')['total'].sum().sort_values(ascending=False)
print('Top Products:')
print(product_revenue.head(10))
```

```python
product_revenue.head(10).plot(kind='bar', figsize=(12,6))
plt.title('Top 10 Products by Revenue')
plt.ylabel('Revenue ($)')
plt.show()
```

```python
category_revenue = df.groupby('category')['total'].sum()
print('Revenue by Category:')
print(category_revenue)
```

```python
category_revenue.plot(kind='pie', autopct='%1.1f%%', figsize=(10,6))
plt.title('Revenue Distribution by Category')
plt.ylabel('')
plt.show()
```

```python
df['date'] = pd.to_datetime(df['date'])
daily_revenue = df.groupby('date')['total'].sum()
daily_revenue.plot(figsize=(14,6), marker='o')
plt.title('Daily Sales Trend')
plt.ylabel('Revenue ($)')
plt.show()
```

```python
customer_value = df.groupby('customer_id')['total'].sum().sort_values(ascending=False)
print('Top 10 Customers:')
print(customer_value.head(10))
```

```python
region_revenue = df.groupby('region')['total'].sum().sort_values(ascending=False)
print('Revenue by Region:')
print(region_revenue)

region_revenue.plot(kind='bar', figsize=(10,6), color=['#FF6B6B', '#4ECDC4', '#45B7D1', '#FFA07A'])
plt.title('Revenue by Region')
plt.ylabel('Revenue ($)')
plt.show()
```

```python
payment_revenue = df.groupby('payment_method')['total'].sum().sort_values(ascending=False)
print('Revenue by Payment Method:')
print(payment_revenue)

payment_revenue.plot(kind='pie', autopct='%1.1f%%', figsize=(10,6), colors=['#FF6B6B', '#4ECDC4', '#45B7D1'])
plt.title('Revenue by Payment Method')
plt.ylabel('')
plt.show()
```

```python
print('\n' + '='*80)
print('EXECUTIVE SUMMARY - E-COMMERCE SALES ANALYSIS')
print('='*80)
print(f'\nTotal Revenue: ${df[\"total\"].sum():,.2f}')
print(f'Total Transactions: {len(df)}')
print(f'Average Order Value: ${df[\"total\"].mean():,.2f}')
print(f'Unique Customers: {df[\"customer_id\"].nunique()}')
print(f'Top Product: {product_revenue.index[0]}')
print(f'Top Region: {region_revenue.index[0]}')
print(f'Preferred Payment: {payment_revenue.index[0]}')
print('\n✅ Analysis Complete!')
print('='*80)
```
